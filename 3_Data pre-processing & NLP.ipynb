```python
import pandas as pd
import re
import string

df = pd.read_csv(r'C:\Users\Soumya_SRB\Desktop\Netflix-Content-Clustering-Engine\NetflixSimple.csv')
# 1. Handle Missing Values 
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')
df['rating'] = df['rating'].fillna('Unknown')


# 2. Function to clean the text
def clean_text(text):
    text = str(text).lower() # Convert to lowercase
    text = re.sub(f'[{re.escape(string.punctuation)}]', '', text) # Remove punctuation
    return text


# 3. Create the "Bag of Words"
# We combine director, cast, genres, and description
df['text_features'] = df['director'] + ' ' + df['cast'] + ' ' + df['listed_in'] + ' ' + df['description']


# 4. Apply the cleaning function
df['text_features'] = df['text_features'].apply(clean_text)

print("--- Phase 2: Data Preprocessing Complete ---")
print(df[['title', 'text_features']].head())
```
