```python
import pandas as pd
import re
import string

# Load raw data
df = pd.read_csv(r'C:\Users\Soumya_SRB\Desktop\Netflix-Content-Clustering-Engine\NetflixSimple.csv')

# Preprocessing
df.fillna('Unknown', inplace=True)
def clean_text(text):
    text = str(text).lower()
    text = re.sub(f'[{re.escape(string.punctuation)}]', '', text)
    return text

df['text_features'] = (df['director'] + ' ' + df['cast'] + ' ' + 
                      df['listed_in'] + ' ' + df['description']).apply(clean_text)

# SAVE: This creates a new file for Step 3 to use
df.to_csv('Netflix_Cleaned.csv', index=False)
print("Step 2 Finished. Created 'Netflix_Cleaned.csv'")
```
